# Introduction:
Covid-19 pandemic has been a global shock and certainly affected multiple markets across the world. This paper will concentrate on the stock market, particularly the newly emerging Electric Vehicles industry companies. Covid-19 caused one of the biggest stock crashes in March 2021, bringing back the volatility levels resembling the Global Financial Crisis in 2008. This paper will focus on brands such as Tesla but also traditional car brands selling EVs. 

## Data:

Data was scraped from the yahoo.finance website. It contains daily historical stock prices of 5 chosen stock companies: Tesla (TSLA), Volkswagen (VOW3), Renault (RNO), Ford (F), and KIA (000270.KS). These are the 5 best-selling EV cars of 2021 in European market


# Conclusion:
The findings showed multiple regressions and other tests performed in order to quantify the relationship between variations in the risk-reward relationship and Covid-19 explanatory factor. In majority of cases a significant relationship was observed between those independent and dependent variables, with confidence intervals of 95% and in some cases even 99%. This research gathered satisfactory evidence supporting our prediction of Covid-19 shock being factored in the riskiness of financial assets in the EV stock market. During Covid- 19, market return coefficient decreased substantially in relation to chosen car companies stock returns. Increase in systemic risk caused by Covid-19 outbreak could have explained an entire market sell-off, rather than an individual stock in March 2021. Some research limitations, however, could exist in selection of the Covid-19 window period, as some could argue that Covid-19 should have been chosen for a longer period than 5 months in our model, as new variants and vaccine responsiveness are still affecting the stock market returns. Other possible errors could have been omitting unmeasurable variables as well not including other industries performances. Investors should remain cautious and perhaps hedge against such occurrences and diversify their portfolios to include a higher proportion of bonds in relation to stocks.
