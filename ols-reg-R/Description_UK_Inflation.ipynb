# Introduction:
The research question of this paper is what economic factors influenced the post Covid crisis inflation in the UK? It is a valid question since many countries are experiencing similar inflationary pressures but to a different degree. Purpose of this paper is to better understand the recent inflation phenomena through analysis of legitimate ONS, fred (and others) data and adaptation of economic theories to design of our model.

# Conclusion:

The negative impact of Covid crisis on the UK economy has proven to be quite large as the world markets remain disturbed 2 years afterwards. Firstly, lockdowns disturbed the aggregate demand as employees were ordered to either work from home or got furloughed due to aggregate supply interruptions. The behaviour of oil markets was quite surprising, firstly oil prices per barrel went into negative territory at the start of crisis bringing lower petrol prices, only so the WTI price per barrel could average approximately $80 in late 2021; and by doing so increasing the costs for business and regular drivers. Depressed demand effect seemed to have surpassed negative aggregate supply shock, since the price level in first quarter of 2020 remained below 2% target at 1.6% - it has then been on a slight downward trend until the second quarter of 2022 when it came up to 1.9%. The last quarter of 2020 yielded record 3.5% quarterly inflation – rate not reached since 1995. Taylor’s rule would imply looking into the futures market to estimate the length of the shock, however in the current market conditions and ongoing Russian-Ukrainian war it could prove misleading. IRFs generated shocks to inflation, brought on inflation responses in directions aligned with economic theory. Perhaps, one could argue that a positive BCI shock should have increased inflation instead of decreasing it, in periods ahead. Fan chart inflation forecasts based on our VAR model implied that non-food/non-energy quarterly inflation would fluctuate around 3-5% level throughout the year. Therefore, in terms of policy recommendations, more aggressive contractionary monetary policy would be advised to Bank of England with gradual but substantial policy rate hikes. From the Fiscal policy perspective stricter fuel duty cuts could decrease costs for businesses and regular consumers, as a 5p cut might not be sufficient. Judging by past inflation crises, opportunity cost of long-run inflation is likely to be higher than further cuts in commodities’ taxation.
